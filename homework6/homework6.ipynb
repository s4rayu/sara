```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(np.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def filter_prime_rows(arr):
    prime_mask = np.array([[is_prime(val) for val in row] for row in arr])
    rows_with_primes = np.any(prime_mask, axis=1)
    return arr[rows_with_primes]

arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

result = filter_prime_rows(arr)
print("Rows with at least one prime:\n", result)
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    
    # odd-numbered rows
    for i in range(0, 8, 2):
        for j in range(0, 8, 2):
            board[i, j] = 1
    
    # even-numbered rows
    for i in range(1, 8, 2):
        for j in range(1, 8, 2):
            board[i, j] = 1
    
    return board

def reverse_checkerboard():
    board = np.ones((8, 8), dtype=int)
    
    # odd
    for i in range(0, 8, 2):
        for j in range(0, 8, 2):
            board[i, j] = 0
    
    # even
    for i in range(1, 8, 2):
        for j in range(1, 8, 2):
            board[i, j] = 0
    
    return board

print(checkerboard())
print(reverse_checkerboard())
```

```python
def expansion(string_array, spaces):
    def expand_string(s, n):
        return (' ' * n).join(s)

    vectorized_expand = np.vectorize(expand_string)
    return vectorized_expand(string_array, spaces)

universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```
