```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)

print(data.shape)

magnitude = data[:,0]
temperature = data[:,1]
# print(magnitude, temperature)

print(np.min(magnitude)) #this is the brightest star
print(np.max(magnitude)) #this is the dimmest star
print(np.min(temperature)) 
print(np.max(temperature))


```

```python
import numpy as np
import matplotlib.pyplot as plt

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8,6))

plt.scatter(log_temperature, log_magnitude, alpha=0.7, s=20, label='Stars')

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

brightest_star_idx = np.argmin(magnitude)  
dimmest_star_idx = np.argmax(magnitude)   
hottest_star_idx = np.argmax(temperature) 
coldest_star_idx = np.argmin(temperature)

critical_stars = [
    (np.argmin(magnitude), "black", "Brightest Star"),
    (np.argmax(magnitude), "purple", "Dimmest Star"),
    (np.argmax(temperature), "green", "Hottest Star"),
    (np.argmin(temperature), "orange", "Coldest Star")
]

for idx, color, label in critical_stars:
    plt.scatter(log_temperature[idx], log_magnitude[idx], color=color, marker="*", s =300, label=label)

plt.legend(fontsize=10)
plt.title("Hertzsprung-Russel Diagram", fontsize=14)
plt.xlabel("Temperature (log K)", fontsize=12)
plt.ylabel("Magnitude", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
