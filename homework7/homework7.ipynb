```python
import numpy as np
import matplotlib.pyplot as plt

def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(25, 5))
line_styles = ['-', '--', '-.', ':', '-']
```

```python
for i, (A, w) in enumerate(zip(amplitude, frequency)):
    linestyle = line_styles[i % len(line_styles)]
```

```python
def plot_sine_waves(x, A, w, linestyle='-'):
    y1 = make_sine_wave(x, A, 1)
    y2 = make_sine_wave(x, 1, w)
    y3 = make_sine_wave(x, A, w)
    
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))
    
    ax1.plot(x, y1, linestyle=linestyle, linewidth=2, label=f'A={A}, ω=1')
    ax2.plot(x, y2, linestyle=linestyle, linewidth=2, label=f'A=1, ω={w}')
    ax3.plot(x, y3, linestyle=linestyle, linewidth=2, label=f'A={A}, ω={w}')
    
    font_size = 14
    
    ax1.set_title('Varying Amplitude', fontsize=font_size+2)
    ax1.set_xlabel('x', fontsize=font_size)
    ax1.set_ylabel('y = A · sin(x)', fontsize=font_size)
    ax1.legend(fontsize=font_size-2)
    ax1.grid(True, alpha=0.3)
    
    ax2.set_title('Varying Frequency Only', fontsize=font_size+2)
    ax2.set_xlabel('x', fontsize=font_size)
    ax2.set_ylabel('y = sin(ωx)', fontsize=font_size)
    ax2.legend(fontsize=font_size-2)
    ax2.grid(True, alpha=0.3)
    
    ax3.set_title('Varying Both Amplitude and Frequency', fontsize=font_size+2)
    ax3.set_xlabel('x', fontsize=font_size)
    ax3.set_ylabel('y = A · sin(ωx)', fontsize=font_size)
    ax3.legend(fontsize=font_size-2)
    ax3.grid(True, alpha=0.3)
    
    plt.tight_layout()
    
    return fig, (ax1, ax2, ax3), (y1, y2, y3)
```

```python
fig, axes, data = plot_sine_waves(x, 2, 3)
```
