```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from astropy import units as u, constants as const
from astropy.units import Quantity
```

```python
df = pd.read_csv("solar_system.csv")

# display(df)

print("---Original---")
print(df.shape, df.shape[0], df.shape[1])

df = df.set_index("Attribute").T #make attribute column the index and transpose the table so that planets = rows and attr = cols
df.index.name = "Planet" #rename index to Planet
df.reset_index(inplace=True) #Move the index back into a column
df.columns.name = None #Remove column index name

print("---Transposed---")
print(df.shape, df.shape[0], df.shape[1])
print(df.columns.tolist())
cols_w_units = [c for c in df.columns if "(" in c and ")" in c]
cols_wo_units = [c for c in df.columns if c not in cols_w_units]
print(len(cols_w_units), cols_w_units)
print(len(cols_wo_units), cols_wo_units)

display(df)

'''
the new shape isnt 11,20 becuase we dont have 11 attr and 20 planets. we have 10 planets (or really objects), and 21 attrs
'''
```

```python
for col in df.columns:
    print(col, df[col].apply(type).unique())

'''
the actual values are of class str, this is problematic because we cannot perform mathematics on them and cannot be plotted/performed operations...etc
'''
```

```python
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df.dtypes

'''
now we have float64-all numerical with decimalcolumns, int64-numerical without decimal points, and object-categorical/string columns.
columns such as planet, global mag field, and ring system, are objects, because they need to be described w actual words
'''
```

```python
display(df)

def attach_units(col, unit, new_col):
    new_values = []
    for val in df[col]:
        new_values.append(val * unit)   
    df[col] = new_values
    df.rename(columns={col: new_col}, inplace=True)

attach_units(
    col="Mass (10^24kg)",
    unit=1e24 * u.kg,
    new_col="Mass (kg)"
)

print(df["Mass (kg)"])
```

```python
attach_units("Perihelion (10^6 km)", 1e7 * u.km, "Perihelion (km)")
attach_units("Aphelion (10^6 km)", 1e7 * u.km, "Aphelion (km)")


smaj = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2


pos = df.columns.get_loc("Aphelion (km)") + 1
df.insert(pos, "Semi-Major Axis (km)", smaj)

print(df.columns)
print(df["Semi-Major Axis (km)"])

display(df)
```
